# Data Generation: Faker & LLM

This notebook demonstrates two approaches to synthetic data generation:

## Faker Generation (no external dependencies)

Generate realistic fake data using built-in templates:
- **person** — names, emails, phone numbers, dates of birth
- **address** — street addresses, cities, states, zip codes
- **company** — company names, catch phrases, industries
- **product** — product names, descriptions, prices, categories

## LLM Generation (requires Ollama)

Generate structured data using a local LLM via Ollama:
- Custom prompts for any domain
- Auto-schema inference from natural language descriptions
- Consistent JSON output

```python
import json
from pathlib import Path

people_file = Path("../output/nb_faker_people.json")

if people_file.exists():
    with open(people_file) as f:
        people = json.load(f)

    print(f"Generated {len(people)} person records")
    print(f"Keys: {list(people[0].keys())}")
    print(f"\nSample records:")
    for record in people[:3]:
        print(json.dumps(record, indent=2))
        print()
else:
    print("nb_faker_people.json not found.")
```

## LLM Generation with Ollama

The following cells require **Ollama** running locally. If Ollama is not available, these cells will fail gracefully.

LLM generation supports:
- **Custom prompts** — describe what data you want in natural language
- **Auto-schema** — describe the structure and the LLM will generate matching JSON

```python
!npm start -- generate -t llm -p "Generate customer support Q&A about product returns" -c 5 -o ../output/nb_llm_qa.json
```

```python
!npm start -- generate -t llm -p "Product reviews" --auto-schema "review with title, body, rating 1-5, pros, cons" -c 5 -o ../output/nb_llm_reviews.json
```

```python
import json
from pathlib import Path

# Inspect LLM Q&A output
qa_file = Path("../output/nb_llm_qa.json")
if qa_file.exists():
    with open(qa_file) as f:
        qa_data = json.load(f)
    print(f"LLM Q&A: {len(qa_data)} records")
    print(f"Keys: {list(qa_data[0].keys())}")
    print(f"\nFirst record:")
    print(json.dumps(qa_data[0], indent=2))
else:
    print("nb_llm_qa.json not found (Ollama may not be running).")

print("\n" + "="*50 + "\n")

# Inspect LLM reviews output
reviews_file = Path("../output/nb_llm_reviews.json")
if reviews_file.exists():
    with open(reviews_file) as f:
        reviews_data = json.load(f)
    print(f"LLM Reviews: {len(reviews_data)} records")
    print(f"Keys: {list(reviews_data[0].keys())}")
    print(f"\nFirst record:")
    print(json.dumps(reviews_data[0], indent=2))
else:
    print("nb_llm_reviews.json not found (Ollama may not be running).")

# Compare structures if both exist
if qa_file.exists() and reviews_file.exists():
    print("\n" + "="*50)
    print("\nStructure comparison:")
    print(f"  Q&A keys:     {list(qa_data[0].keys())}")
    print(f"  Reviews keys: {list(reviews_data[0].keys())}")
    print(f"\nThe auto-schema option produces structured fields matching your description,")
    print(f"while free-form prompts let the LLM decide the output structure.")
```

## Combining Generation with Formatting

Generated data can be piped through the same `format` command used in the quickstart notebook to produce ML-ready datasets. For more advanced workflows that chain multiple steps together, check out the **pipeline** feature.

Example pipeline idea:
1. Generate 100 Faker person records
2. Generate LLM-powered biographical summaries for each
3. Format into ShareGPT conversation format
4. Export with train/test splits

```python
import os
from pathlib import Path

# Cleanup: remove nb_ prefixed files from output/
output_dir = Path("../output")
removed = []

for item in output_dir.iterdir():
    if item.name.startswith("nb_"):
        if item.is_file():
            item.unlink()
            removed.append(item.name)
        elif item.is_dir():
            import shutil
            shutil.rmtree(item)
            removed.append(item.name + "/")

if removed:
    print(f"Cleaned up {len(removed)} item(s):")
    for name in removed:
        print(f"  - {name}")
else:
    print("Nothing to clean up.")
```
